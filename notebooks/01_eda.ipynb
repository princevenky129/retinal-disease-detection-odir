# Phase 1: EDA — ODIR-5K
Class distribution, age/sex stats, multi-label co-occurrence, visual sample inspection, left/right pairing check.

```python
import sys, os
sys.path.append(os.path.abspath('..'))
import pandas as pd
import matplotlib.pyplot as plt
import yaml

with open('../config/config.yaml') as f:
    cfg = yaml.safe_load(f)

raw_df = pd.read_csv(cfg['data']['raw_annotations_csv'])
print(raw_df.shape)
raw_df.head()
```

```python
# Confirm actual column names — adjust scripts/prepare_data.py if these differ
raw_df.columns.tolist()
```

```python
CLASSES = ['N','D','G','C','A','H','M','O']
counts = raw_df[CLASSES].sum().sort_values(ascending=False)
counts.plot(kind='bar', title='Class counts (patient-level)')
plt.show()
counts
```

```python
# Multi-label co-occurrence: how many patients have >1 disease label
num_labels = raw_df[CLASSES].sum(axis=1)
num_labels.value_counts().sort_index()
```
